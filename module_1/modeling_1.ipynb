```python
import sympy
from sympy.plotting import plot  # Функция для отображения графиков
```

```python
U, T_m, t = sympy.symbols(
    'U, T_m, t')  # обозначение символов переменных напряжения, постоянной времени и самого времени
k_e = sympy.symbols('k_e',
                    positive=True)  # конструктивная постоянная, должна быть положительной для однозначного решения уравнения
omega = sympy.Function('omega')  # обозначение функции угловой скорости
```

```python
ode = sympy.Eq(omega(t).diff(t), U / (k_e * T_m) - omega(t) / T_m)  # задание дифференциального уравнения двигателя
ode
```

```python
ode_sol = sympy.dsolve(ode, omega(t), ics={omega(0): 0})  # решение дифференциального уравнения
ode_sol
```

```python
ode_sol = sympy.expand(ode_sol)  # упрощение уравнения (не упростили, по итогу)
ode_sol
```

```python
ode_num = ode_sol.subs({k_e: 0.5, T_m: 0.08, U: 8})  # подстановка числовых значений в уравнение
ode_num
```

```python
# from sympy.plotting import plot                                 # импорт функции изображения графиков
plot(ode_num.rhs, (t, 0, 0.5), line_color='red',
     title='Разгон двигателя', xlabel='time,[sec]',
     ylabel='speed, [rad/sec]')  # изображение графика с различными настройками
```

```python
theta_eq = sympy.integrate(ode_sol.rhs, (t, 0, t))  # интегрирование уравнения для получения уравнения углового смещения
theta_eq
```

```python
theta = sympy.Function('theta')  # обозначение функции углового смещения
ode_theta = sympy.Eq(theta(t).diff(t, 2),
                     -theta(t).diff(t) / T_m + U / (T_m * k_e))  # дифференциальное уравнение углового смещения
ode_theta
```

```python
ode_sol_theta = sympy.dsolve(ode_theta, theta(t), ics={theta(0): 0, theta(t).diff(t).subs(t, 0): 0})  # решение дифференциального уравнения
ode_sol_theta
```

```python
theta_num = ode_sol_theta.subs({k_e: 0.5, T_m: 0.08, U: 8})  # подстановка числовых значений в уравнение
theta_num
```

```python
plot(theta_num.rhs, (t, 0, 0.5), line_color='red',
     title='Разгон двигателя', xlabel='time,[sec]',
     ylabel='angle, [rad]')  # изображение графика с различными настройками
```
